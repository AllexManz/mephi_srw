# Stage comparison

Compare baseline, pretrain and RL reports produced by the experiment runner.

```python
import json
from pathlib import Path
import pandas as pd

root = Path('../outputs/experiments/security-llm')
rows = []
for path in root.glob('*.json'):
    if path.name == 'manifest.json':
        continue
    data = json.loads(path.read_text(encoding='utf-8'))
    for metric, value in data.get('metrics', {}).items():
        rows.append({'file': path.name, 'model': data.get('model_id'), 'stage': data.get('stage'), 'metric': metric, 'value': value})

if rows:
    display(pd.DataFrame(rows).pivot_table(index=['model', 'stage'], columns='metric', values='value'))
else:
    print('No benchmark reports found')
```
