# Data overview

Inspect local train/eval and external benchmark composition without running training.

```python
import json
from collections import Counter
from pathlib import Path

def load(path):
    return json.loads(Path(path).read_text(encoding='utf-8'))

train = load('../data/processed/train.json')
eval_data = load('../data/processed/eval.json')
print('train:', len(train), 'eval:', len(eval_data))
print('train fields:', sorted(train[0]))
print('threat levels:', Counter(x.get('threat_level', 'missing') for x in train))
```

```python
external_path = Path('../data/external/external_benchmark_1500.json')
if external_path.exists():
    external = load(external_path)
    print('external:', len(external))
    print(Counter(x.get('source') for x in external))
else:
    print('Run src/evaluation/prepare_external_dataset.py first')
```
